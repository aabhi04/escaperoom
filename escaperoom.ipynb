```python
import random

# Global Inventory System to store collected clues [cite: 43, 57]
inventory = []

def collect_item(item):
    """Adds a newly discovered item/clue to the player's inventory list[cite: 44]."""
    if item not in inventory:
        inventory.append(item)
        print(f"\n[SYSTEM] You picked up: {item}!")
    else:
        print("\nYou already have this item.")

def show_inventory():
    """Prints all items currently held by the player[cite: 44]."""
    print("\n--- INVENTORY ---")
    if not inventory:
        print("(Your inventory is empty)")
    else:
        for item in inventory:
            print(f"- {item}")
    print("-----------------")

def check_answer(user_input, correct_answer):
    """Compares player input with the actual puzzle answer[cite: 42, 58]."""
    # .strip() removes accidental extra spaces, .lower() ignores letter casing
    return user_input.strip().lower() == str(correct_answer).strip().lower()

# ==================== ROOM MANAGERS ====================

def room_one():
    """Room 1: Uses a Number Guessing Puzzle[cite: 40, 48]."""
    print("\n==================================================")
    print("ROOM 1: THE DIGITAL LAB")
    print("The heavy steel door locks behind you. A computer terminal flashes.")
    print("Puzzle: Guess the hidden binary key variable (a number between 1 and 5).")
    print("==================================================")

    secret_number = random.randint(1, 5) # Generates a random puzzle target [cite: 63]

    # Loop keeps player stuck until they get it right [cite: 55]
    while True:
        guess = input("\nEnter your number guess (or type 'inv' to view inventory): ")

        if guess.lower() == 'inv':
            show_inventory()
            continue

        if check_answer(guess, secret_number): # [cite: 42]
            print("\nCorrect! The terminal buzzes open a small side drawer.")
            collect_item("Rusty Key") # Collect a mandatory item for room transition [cite: 44]
            break
        else:
            print("Incorrect signal. The terminal flashes red. Try again!")

    print("\nYou unlock the side door with the Rusty Key and step through...")
    room_two() # Transition to next room

def room_two():
    """Room 2: Uses a Riddle / Word Puzzle[cite: 40, 50, 51]."""
    print("\n==================================================")
    print("ROOM 2: THE ANCIENT LIBRARY")
    print("Dusty bookshelves line the walls. A stone statue stands in the center.")
    print("Riddle: 'The more of them you take, the more you leave behind. What am I?'")
    print("==================================================")

    correct_solution = "footsteps"

    while True:
        guess = input("\nEnter your answer (or type 'inv'): ")

        if guess.lower() == 'inv':
            show_inventory()
            continue

        if check_answer(guess, correct_solution): # [cite: 42]
            print("\nThe statue slides back, revealing a shining metallic fragment.")
            collect_item("Cipher Code: 99") # Collect raw data clue [cite: 44, 52]
            break
        else:
            print("The statue remains completely motionless. That is not the answer.")

    print("\nA trap door on the floor opens up. You climb down into the depths...")
    room_three() # Transition to next room

def room_three():
    """Room 3: Uses a Math Puzzle[cite: 40, 49]."""
    print("\n==================================================")
    print("ROOM 3: THE BOILER CORE")
    print("Steam blocks your path. A pressure valve is locked behind a numeric keypad.")
    print("Math Puzzle: Solve the sequence multiplier: 2, 4, 8, 16, ?")
    print("==================================================")

    correct_solution = 32

    while True:
        guess = input("\nEnter the next number in the sequence (or type 'inv'): ")

        if guess.lower() == 'inv':
            show_inventory()
            continue

        if check_answer(guess, correct_solution): # [cite: 42]
            print("\nThe steam clears! You find a security card sitting on the control desk.")
            collect_item("Magnetic Card") # Collect clue [cite: 44]
            break
        else:
            print("Pressure rising! Incorrect sequence value. Try again.")

    print("\nYou approach the final bulkhead vault exit...")
    final_escape() # Move to final validation gate [cite: 46]

def final_escape():
    """Final Gate: Validates collected inventory tokens[cite: 45]."""
    print("\n==================================================")
    print("FINAL ESCAPE BULKHEAD")
    print("You have reached the main surface exit. The console requires items to unlock.")
    print("==================================================")

    input("Press Enter to insert your collected keycards and run items verification...")
    show_inventory()

    # Conditional logic checks if inventory holds all essential targets [cite: 56, 57]
    if "Rusty Key" in inventory and "Cipher Code: 99" in inventory and "Magnetic Card" in inventory:
        print("\n[SUCCESS] The system processes your tokens...")
        print("The massive outer gate swings wide open! Daylight streams in.")
        print("CONGRATULATIONS! YOU HAVE ESCAPE THE MYSTERIOUS BUILDING! [cite: 66]")
    else:
        print("\n[FAILED] The alarm blares. You do not possess the required inventory clues.")
        print("You are trapped in the facility forever...")

def start_game():
    """Main execution point initializing variables and text intro[cite: 38]."""
    global inventory
    inventory = [] # Resets items pool on new runtime init

    print("************************************************")
    print("       WELCOME TO THE ESCAPE ROOM PUZZLE        ")
    print("************************************************")
    print("Instructions: Solve puzzles to unlock your way forward. [cite: 30]")
    print("Type 'inv' at any prompt to inspect your current inventory keys. [cite: 44]")
    input("\nPress Enter to begin your survival challenge...")

    room_one() # Kick off initial state sequence

# --- Execute Game Launch Trigger ---
if __name__ == "__main__":
    start_game()
```
